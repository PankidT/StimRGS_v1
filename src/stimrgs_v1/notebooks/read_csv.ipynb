```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
path_1 = "data_rgs_plots_bsm_0.5_3hops/rgs_summary.csv"
df_1 = pd.read_csv(path_1)

path_2 = "data_rgs_plots_bsm_0.5/rgs_summary.csv"
df_2 = pd.read_csv(path_2)

path_3 = "data_rgs_plots_bsm_0.5_4hops/rgs_summary.csv"
df_3 = pd.read_csv(path_3)

arms = 12

filtered_df_1 = df_1[df_1['arm_size'] == arms]
filtered_df_2 = df_2[df_2['arm_size'] == arms]
filtered_df_3 = df_3[df_3['arm_size'] == arms]

plt.figure(figsize=(10, 5))
plt.bar(filtered_df_2['edge_count'], filtered_df_2['mean_bell_pairs'], color='green', alpha=1.0, label="2 Hops")
plt.bar(filtered_df_1['edge_count'], filtered_df_1['mean_bell_pairs'], color='blue', alpha=1.0, label='3 Hops')
plt.bar(filtered_df_3['edge_count'], filtered_df_3['mean_bell_pairs'], color='red', alpha=1.0, label="4 Hops")

# Add labels and title for better visualization
plt.xlabel('Edge Count')
plt.ylabel('Mean Bell Pairs')
plt.title(f'Mean Bell Pairs vs Edge Count (Arm Size = {arms})')
plt.legend()
# Display the plot
plt.show()
```

```python
path_1 = "data_rgs_plots_bsm_0.5_3hops/rgs_summary.csv"
df_1 = pd.read_csv(path_1)

path_2 = "data_rgs_plots_bsm_0.5/rgs_summary.csv"
df_2 = pd.read_csv(path_2)

for i in range(1, 13):
    arms = i

    filtered_df_1 = df_1[df_1['arm_size'] == arms]
    filtered_df_2 = df_2[df_2['arm_size'] == arms]

    # Create the figure and axis
    fig, ax = plt.subplots(figsize=(10, 6))

    # Plot first dataset normally (positive y-axis)
    ax.bar(filtered_df_1['edge_count'], filtered_df_1['mean_bell_pairs'], 
        color='blue', alpha=0.5, label='3 Hops')

    # Plot second dataset with negative values (negative y-axis)
    ax.bar(filtered_df_2['edge_count'], -filtered_df_2['mean_bell_pairs'], 
        color='red', alpha=0.5, label='2 Hops')

    # Add a horizontal line at y=0
    ax.axhline(y=0, color='black', linestyle='-', alpha=0.3)

    # Add labels and title
    ax.set_xlabel('Edge Count')
    ax.set_ylabel('Mean Bell Pairs')
    ax.set_title(f'Mean Bell Pairs vs Edge Count (Arm Size = {arms})')

    # Customize y-axis labels to show absolute values
    y_ticks = ax.get_yticks()
    ax.set_yticklabels([str(abs(y)) for y in y_ticks])

    # Add legend
    ax.legend()

    # Show plot
    plt.tight_layout()
    plt.show()
```
